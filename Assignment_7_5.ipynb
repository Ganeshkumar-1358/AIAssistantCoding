TASK - 1  (Mutable Default Argument – Function Bug)

Corrected Code:

```python
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(add_item(1))
print(add_item(2))
```

TASK - 2  (Floating-Point Precision Error)

```python
def check_sum():
    return abs((0.1 + 0.2) - 0.3) < 1e-9

print(check_sum())
```

TASK -3 (Recursion Error – Missing Base Case)

Corrected Code:

```python
def countdown(n):
    if n <= 0:        # Base case (stop condition)
        print("Done!")
        return
    print(n)
    countdown(n - 1)

countdown(5)
```
